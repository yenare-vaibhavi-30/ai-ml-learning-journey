```python
import numpy as np
```

```python
 # 1. Create a NumPy array
arr = np.array([1,2,3,4,5,6])
print(arr, arr.size)
```

```python
#2. Find array dimensions

arr = np.array([[1,2,3,4,5,6], [2,4,3,4,5,3]])
print("Array", arr)
print("Dimensions : ", arr.ndim )
```

```python
#Find shape and size
arr = np.array([[1,2,4,6,5],[2,4,6,4,2]])
print(arr.shape, arr.size)
```

```python
#Find data type
arr = np.array([1,2,4,6,5])
print(arr.dtype)
```

```python
#Create zeros and ones
zero_arr = np.zeros((3,4))
print("Zero Array")
print(zero_arr)

ones_arr = np.ones((2,3))
print("Once Array")
print(ones_arr)
```

```python
#6. Generate a range of numbers

num = np.array(range(1,11))
print(num)
```

```python
# 7. Create evenly spaced numbers
arr = np.linspace(0,1,5)
print(arr)
```

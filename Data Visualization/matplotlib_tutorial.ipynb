```python
# Real data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

# plot
plt.plot(years, oscar_revenue)

plt.title("Oscar Movies Revenue in each Year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
```

```python
# Non-Oscar Winner Movies

non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [1005, 170, 427, 133, 232] # in $M

plt.plot(years, oscar_revenue, label="Oscar Movies")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")
plt.title("Oscar Movies vs Non_Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend()
```

```python
oscar_revenue = [1005, 170, 427, 133, 232]

non_oscar_revenue = [1005, 170, 427, 133, 232]

plt.plot(years, oscar_revenue,
         label="Oscar Movies",
         color="blue",
         linestyle="-",
         marker="o")

plt.plot(years, non_oscar_revenue,
         label="Non-Oscar Movies",
         color="orange",
         linestyle="--",
         marker="*")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

plt.legend()
plt.show()
```

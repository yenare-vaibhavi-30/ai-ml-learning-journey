```python
# Real data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

# plot
plt.plot(years, oscar_revenue)

plt.title("Oscar Movies Revenue in each Year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
```

```python
# Oscar Winner Movies
oscar_revenue = [1005, 170, 427, 133, 232]

# Non-Oscar Winner Movies
non_oscar_movies = [
    "Slumdog Millionaire",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]

non_oscar_revenue = [378, 2788, 822, 185, 275]

plt.plot(years, oscar_revenue, color = "red",marker = "<",linestyle = "-.",linewidth = 5, label="Oscar Movies")

plt.plot(years, non_oscar_revenue,  label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.show()
```

```python
# Bar Charts

movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
movies_revenue = [1005, 170, 427, 133, 232] 

plt.bar(years, movies_revenue)
plt.title("Revenue in each Year")
plt.xlabel("Year")
plt.ylabel("Revenue($M)")
for i in range(len(years)):
    plt.text(years[i],movies_revenue[i] + 10, str(movies_revenue[i]), ha="center")
```

```python
import numpy as np
import matplotlib.pyplot as plt

oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

oscar_revenue = [1005, 170, 427, 133, 232]

non_oscar_movies = [
    "Slumdog Millionaire",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]

non_oscar_revenue = [378, 2788, 822, 185, 275]

years = [2008, 2009, 2010, 2011, 2012]

x = np.arange(len(years))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label="Oscar Movies", color= "yellow")
plt.bar(x+width/2, non_oscar_revenue, width, label="Non Oscar Movies")

plt.title("Oscar vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue ($M)")

plt.xticks(x, years)
plt.legend()
plt.show()
```

```python
# Scatter Plots

people = ["Person A", "Person B", "Person C", "Person D", "Person E",
           "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [26, 43, 86, 35, 87, 23, 64, 75, 57, 65]
bp = [130, 103, 158, 123, 134, 135, 153, 129, 145, 130]
#colors = ["green" if x < 135 else "red" for x in bp ] 
#plt.scatter(age, bp, color= colors, s = bp, marker = ">", alpha= 0.6)
plt.scatter(age, bp, s = bp, cmap = "OrRd", c = bp)
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.colorbar(label="BP")

for i in range(len(people)):
    plt.annotate(people[i], xy = (age[i], bp[i]), xytext=(age[i], bp[i]))
```

```python
#pie charts

expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amount = [100, 143, 400, 153, 120]

#colors = ["red", "black", "pink", "blue", "yellow"]

#plt.style.use("classic")

explode = [0, 0, 0, 0.2, 0]
plt.pie(amount, labels= expenses, autopct="%1.1f%%",
       wedgeprops = {
           "edgecolor": "black",
           "linewidth": 1,
           "linestyle": "--",
           "alpha": 0.6,
           #"fill": False
       },
        explode=explode,
        startangle=90,
        shadow=True
       )
#plt.style.available
plt.show()
```

```python
# Histogram

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

scores = np.random.normal(70, 10, 100)  # normal distribution
bins = [0,30,70,90,100]
plt.hist(scores, color="lightblue", edgecolor="black", bins = bins)
plt.ylabel("# of students")
plt.xlabel("scores")
print(scores)
```

```python
marks = [
    25, 64, 23, 86, 95, 35, 76, 54, 65, 34,
    76, 43, 75, 65, 34, 76, 34, 76, 54, 76,
    54, 87, 23, 87, 34, 76, 36, 87, 98, 26
]

plt.hist(marks, color="lightblue", edgecolor="black", bins = bins)

plt.axvline(
    x = 33,
    label="Passing marks (33)",
    color="red",
    linewidth="2",
    linestyle="--"
)
plt.ylabel("# of students")
plt.xlabel("scores")
plt.title("distribution of marks for students")
```

```python
# Multiple datasets

legit_transactions = [
    5.99, 10.49, 25.50, 49.99, 75.00,
    99.99, 1200.50, 1050.00, 199.99, 2049.99,
    299.50, 350.00, 4500.00, 500.00, 650.00,
    750.00, 899.99, 1000.00, 1250.00, 1500.00,
    1750.00, 2000.00, 2500.00, 3000.00
]

fraud_transactions = [
    50, 1100, 200, 7000, 750,
    25000, 3000, 350, 4000, 4500,
    5000, 600, 5000, 8000, 1000
]

plt.hist(fraud_transactions, color="red", bins=10, label="Fraud", edgecolor="black", alpha=0.5)
plt.hist(legit_transactions, color="green", bins=10, label="Legit", edgecolor="black", alpha=0.5)
plt.ylabel("Amount (in $)")
plt.xlabel("Frequency")
plt.title("Compare legitimate and fraud transactions.")
plt.legend()
```

```python
#Assignment Problems
#. For the following dataset generated using NumPy:Q1
# np.random.normal(loc=70, scale=10, size=200)
#• Plot a of scoreshistogram
#• Draw a vertical line for the mean & try to interpret the results.

    
import numpy as np
import matplotlib.pyplot as plt

score = np.random.normal(loc=70, scale=10, size=200)
mean_score = np.mean(score)
plt.hist(score, bins=10, edgecolor="black")

plt.axvline(mean_score, color="red", linestyle="--",
            linewidth=2, label=f"Mean = {mean_score:.2f}")

plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.title("Histogram of Students scores")

plt.legend()
plt.grid(axis="y", alpha=0.3)
```

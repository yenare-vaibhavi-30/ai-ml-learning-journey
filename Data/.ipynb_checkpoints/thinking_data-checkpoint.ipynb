```python
import json
```

```python
# load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

        return data
        
```

```python
load_data("store_data.json")
```

```python
#clean and structure data
def clean_data(data):
    text_to_num = {
        "one": 1, "two": 2, "three": 3,
        "four": 4, "five": 5
    }

    for user in data:
        # clean ratings
        raw_rating = user["rating"].lower()

        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]

        user["rating"] = raw_rating
```

```python
clean_data(data)
print(data)
```

```python
#AQI Data set
df = pd.read_csv("globalAirQuality.csv")
df
```

```python
#Select

#df["city"]
#df["country"]
df[["city", "aqi"]]
```

```python
#select row (label and index based) - loc & iloc
df.loc[0]  #label   accsesing coloum data 
df.loc[0:2]  #we pase the starting index : ending index

df.iloc[2]  #index    accessing row data
df.iloc[0:4]

#cell - row, col
df.loc[0:2, ["city","aqi","latitude","country"]]  #label  #row
df.iloc[0:3, 2 ]  #index  #coloum

#cell
df.at[0,"city"]
df.iat[0,2]

cities = df["city"].copy()
cities[0] = "Mumbai"
cities
```

```python
#filtering of data

df[ df["aqi"] > 100]

df[ (df["aqi"] > 100) & (df["temperature"] > 30)]

df [ df["latitude"] > 30]

df [df["city"] == "New York"]

api_data = df[ (df["aqi"] > 100) & (df ["temperature"] > 30)  ][["city","aqi"]]
api_data

api_data.iloc[0]
api_data.loc[6]

#query
api_val = 100
df.query("aqi > 100 & temperature > 30")[["city","aqi"]]
#df.query("aqi > @aqi_val & temperature > 30")[["city", "aqi"]]
```

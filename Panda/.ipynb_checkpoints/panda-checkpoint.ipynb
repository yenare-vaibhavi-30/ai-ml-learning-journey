```python
#cleaning data missing values

df = pd.read_csv("student_lifestyle_100k.csv")

#handles duplicate

df["Department"].duplicated()
df2 = df.copy()
df.drop_duplicates(inplace=True)
```

```python
# data type

df.dtypes
df2 = df.copy()
df2 = df2.fillna(0)
df2 = df2["Age"].astype("int64").copy()
df2.dtypes
```

```python
df
```

```python
#feature engineering

df2 = df.copy()
df2["Physical_Activity"] = df2["Social_Media_Hours"].apply(lambda x : "20%" if x >= 60000 else "10%")

gender_map = {"Male" : "M", "Female" : "F", "Unknown" : "U"}
df2["Gender"] = df2["Gender"].map(gender_map)

#df2 = df2.assign(new_income = df2["Social_Media_Hours"] = 1.1)

df2["Department"] = df2["Department"].replace("Science", "Math")

df2.columns = ["Student_ID","Age","Gender","Department","CGPA","Sleep_Duration","Study_Hours","Social_Media_Hours","Physical_Activity","Stress_Level","Depression"]
df2.rename(columns={"CGPA":"Percentage"})
df2.rename(index={1:"First"})
df2.sort_values("Sleep_Duration")
df2.sort_values(["Sleep_Duration","Age"])

#Reset
sorted_df = df.sort_index()
sorted_df.reset_index()

#Ranking 
sorted_df["Ranking"] = sorted_df["Stress_Level"].rank(ascending=False,method="dense")
sorted_df
```

```python
# Melt and Pivot

df = pd.DataFrame({
    "country": ["USA","USA","India","india"],
    "year" : [2020, 2021, 2020, 2021],
    "sales" : [100, 130, 90, 22],
    "profit":[23,26,20,10],
    "tax": [12,45,65,87]
})

melted_df = df.melt(
    id_vars = ["country","year"],
    value_vars = ["sales", "profit"],
    var_name = "metrics",
    value_name = "value"
)

melted_df.pivot(
    index = ["country","year"],
    columns = "metrics",
    values = "value"
)
```

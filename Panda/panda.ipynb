```python
import pandas as pd
```

```python
info = {
    "Name" : ["Vaibhavi", "Siddu", "Vaishnavi", "Madhu"],
    "CGPA" : [9.3, 9.5,8.9,9.9]
}
df = pd.DataFrame(info)
print(df)
```

```python
series = pd.Series([1,2,3,4,5])
print(series)
print(type(series))

#index
print(series[0])
print(series[2])
```

```python
series = pd.Series([26, 54, 23, 53], index= ["Adam", "Eve", "Bob", "Charlie"])
print(series)

print(series["Adam"])
print(series["Bob"])
print(series[0])
print(series.index)
```

```python
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])

print(s1 + s2)
s1[0] = 100
changed_s1 = s1.drop(0)
print(s1)
print(changed_s1)
```
